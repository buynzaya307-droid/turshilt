###Create a dict

```python
weeks = {
    "1" : "Monday",
    "2" : "Tuesday",
    "3" : "Wednesday",
    "4" : "Thursday",
    "5" : "Friday",
    "6" : "Saturday",
    "7" : "Sunday",
}
print(weeks)
```

###Key


```python
print(weeks['1'])
```

```python
print(weeks['3'])
```

###Get()

```python
print(weeks.get("9","Гараг нь зөвхөн 7 өдрөөс бүрдэнэ!"))
```

###Change Value

```python
weeks['1'] = "Mon"
```

```python
print(weeks)
```

###Loop througth a dict

###Key

```python
for x in weeks:
 print(x)
```

```python
for x in weeks.values():
  print(x)
```

###Items

```python
for x, y in weeks.items():
 print(x, y)
```

###Сарыг тоо руу шилжүүлдэг

```python
months = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}
month_name = "March"
if month_name in months:
    print(f"{month_name} сар бол: {months[month_name]}-р сар")
else:
    print("Ийм сар олдсонгүй")
```

###Англи үгийг монгол үг рүү орчуулах

```python
dictionary = {
    "apple": "алим",
    "book": "ном",
    "car": "машин",
    "dog": "нохой",
    "school": "сургууль"
}
word = "book"
if word in dictionary:
    print(f"'{word}' гэдэг үг нь монголоор: {dictionary[word]}")
else:
    print("Энэ үг толь бичигт байхгүй байна")
```
